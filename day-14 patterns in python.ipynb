```python
n=int(input())
num=1
for i in range(n):
    for j in range(n):
        print(num,end=" ")
        num+=1
    print()
```

```python
n=4
for i in range(n):
    for j in range(n):
        if i %2==0:
            print("+",end=" ")
        else:
            print("-",end=" ")
    print()
```

```python
n=5
k=1
for i in range(n+1):
    for j in range(i):
        if i>=j:
            print(k,end=" ")
            k+=1
    print()
```
