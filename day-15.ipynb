 Problem Solving
 - find length of number eg: 456=3
 -  sum of digit in a number eg:123=6
 -  reverse a number eg:321=123
 - armstrong number 

```python
#find length of a number
n=1234
count=0
while n>0:
    n=n//10
    count+=1
print(count)
```

```python
#find sum of number
n=int(input())
s=0
while n>0:
    digit=n%10
    n=n//10
    s=s+digit
print(s)
```

```python
#Reverse of a number
n=int(input())
s=0
while n>0:
    digit =n%10
    n=n//10
    s=s*10+digit
print(s)
```
